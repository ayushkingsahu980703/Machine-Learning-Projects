## 4. Feature Selection and Scaling


```python
features=['Inns', 'NO', 'Runs', 'HS', 'Ave', 'SR', '100',
       '50', '0', 'experience']
```

```python
x= df[features].copy()
```

```python
x.head(10)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_scaled=sc.fit_transform(x)
```

```python
x_scaled=pd.DataFrame(x_scaled,columns=x.columns,index=x.index)
x_scaled
```

## 5. K-Means Model Selection


```python
from sklearn.cluster import KMeans

k_value=[2,3,4,5,6,7,8,9,10]
wcss=[]
for k in k_value:
  model=KMeans(n_clusters=k,
               init='k-means++',
               n_init=10,
               max_iter=100,
               random_state=42
  )
  model.fit(x_scaled)
  wcss.append(model.inertia_)
```

```python
fig_elbow = px.line(
    x=k_value,
    y=wcss,
    markers=True,
    title='Elbow Method: WCSS vs Number of Clusters',
    labels={'x': 'Number of Clusters (K)', 'y': 'WCSS'}
)

fig_elbow.update_traces(
    marker=dict(size=9),
    line=dict(width=3)
)

fig_elbow.show()
```

### Silhouette Score


```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in k_value:
  model=KMeans(
      n_clusters=k,
      init='k-means++',
      n_init=10,
      max_iter=100,
      random_state=42
  )
  labels= model.fit_predict(x_scaled)
  score=silhouette_score(x_scaled, labels)
  silhouette_scores.append(score)
```

```python
fig_silhouette = px.line(
    x=k_value,
    y=silhouette_scores,
    markers=True,
    title='Silhouette Score vs Number of Clusters',
    labels={'x': 'Number of Clusters (K)', 'y': 'Silhouette Score'}
)

fig_silhouette.update_traces(
    marker=dict(size=9),
    line=dict(width=3)
)

fig_silhouette.show()
```

## 6. Final K-Means Model


```python
optimal_k = 2
final_model=KMeans(
    n_clusters=optimal_k,
    init='k-means++',
    max_iter=100,
    n_init=10,
    random_state=42

)
cluster_labels=final_model.fit_predict(x_scaled)
```

### Cluster Assignment


```python
df['Cluster'] = cluster_labels
```

## 10. Final Results and Portfolio Interpretation

### Model Selection

The Elbow Method and Silhouette Score were used together to evaluate candidate values of K. The highest tested Silhouette Score occurred at **K = 2 (approximately 0.261)**, while the WCSS curve showed progressively smaller reductions as K increased. Based on these diagnostics, **K = 2** was selected for the final model.

### Cluster Size

The final model assigns **52 players to Cluster 0** and **27 players to Cluster 1**.

### Cluster 0 — comparatively lower career-volume profile

Cluster 0 has lower average matches, innings, runs, balls faced, centuries, half-centuries and experience than Cluster 1. It should not be interpreted as a group of low-quality players: individual players within this cluster can still have strong batting averages or strike rates.

### Cluster 1 — higher career-volume and longer-career profile

Cluster 1 has substantially higher average career volume. Its mean career runs are approximately **10,413**, compared with approximately **6,167** for Cluster 0. It also has higher averages for matches, innings, balls faced, centuries, half-centuries and experience.

The mean batting averages are relatively close (**39.27 vs 38.13**), so the clustering should not be described as a simple 'better vs worse players' classification. The strongest distinction is associated with **career volume, longevity and accumulated scoring output**.

### PCA Interpretation

The first two principal components explain **61% of the variance** in the standardized feature set. PC1 has its largest positive loadings on Runs, Innings, 50s, 100s and Experience, making it primarily associated with career volume and scoring milestones. PC2 has its largest positive loadings on Average, Highest Score and 100s, with Strike Rate also contributing positively.

### Key Findings

- K-Means identified two broad statistical profiles in the 79-player dataset.
- Cluster 1 is characterized by higher career volume and longevity.
- Cluster 0 contains more players and has lower average career volume, but includes players with strong individual batting statistics.
- Career volume contributes strongly to the separation between clusters.
- PCA provides a two-dimensional representation of the data, with the first two components capturing 61% of total variance.


## 11. Limitations

- The dataset contains **79 players**, so the results should not be generalized to all cricket players without additional data.
- K-Means requires a predefined number of clusters; K = 2 was selected using the Elbow Method and Silhouette Score, but this does not prove that two clusters are the only meaningful representation.
- The maximum Silhouette Score is approximately **0.261**, so the clusters should be interpreted as statistical groupings rather than perfectly separated player types.
- The analysis uses career-level batting statistics and does not account for match context, opposition, playing position, era effects or other contextual variables.
- Cluster labels are identifiers, not quality rankings.


## 12. Conclusion

This project demonstrates how unsupervised machine learning can be used to discover patterns in cricket batting careers without a predefined target variable. The K-Means model produced two broad statistical profiles, with the strongest differences related to career volume, scoring output and longevity.

The analysis also shows why clustering should be interpreted carefully: batting average and strike rate contribute to the player profiles, but the separation is not simply a measure of player quality.

### Portfolio Takeaway

> **K-Means clustering was used to segment cricket players based on career batting statistics. Elbow and Silhouette analysis supported K = 2, while cluster profiling showed that the strongest distinction between the groups was career volume and longevity rather than batting average alone.**
